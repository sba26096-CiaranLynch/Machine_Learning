## CA1 Machine Learning

### Banking Dataset taked from UC Irvine Machine Learning Repository
####  https://archive.ics.uci.edu/dataset/2/adult
#### creators Barry Becker & Ronny Kohavi
#### Predict whether annual income of an individual exceeds $50K/yr based on census data. Also known as "Census Income" dataset.


#### Associated Task = Classification
#### Dataset Characteristics = Multivariate
#### Extraction was done by Barry Becker from the 1994 Census database.  A set of reasonably clean records was extracted using the following conditions: ((AAGE>16) && (AGI>100) && (AFNLWGT>1)&& (HRSWK>0))
#### Prediction task is to determine whether a person's income is over $50,000 a year.

### Add a dictionary to place columns into the headers 

```python
Column_Header =  {
                    df.columns[0]:  'age',
                    df.columns[1]:  'workclass',
                    df.columns[2]:  'fnlwgt',
                    df.columns[3]:  'education',
                    df.columns[4]:  'education-num',
                    df.columns[5]:  'marital-status',
                    df.columns[6]:  'occupation',
                    df.columns[7]:  'relationship',
                    df.columns[8]:  'race',
                    df.columns[9]:  'sex',
                    df.columns[10]:  'capital-gain',
                    df.columns[11]: 'capital-loss',
                    df.columns[12]: 'hours-per-week',
                    df.columns[13]: 'native-country',
                    df.columns[14]: 'income',
                    
                  }


df.rename(columns = Column_Header, inplace = True) # adding a dictionary to place columns into the headers 
```

```python
# create duplicate dataframe with required columns only
df_income = df[
                ["age", 
                 "education", 
                  "education-num", 
                  "marital-status",
                  "occupation", 
                  "relationship",
                  "capital-gain", 
                  "capital-loss", 
                  "hours-per-week", 
                  "income"] 
                ].copy()
```
